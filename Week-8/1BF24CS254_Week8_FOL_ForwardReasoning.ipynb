```python

# Forward Chaining in First Order Logic

# Step 1: Display Problem Statement
print("PROBLEM STATEMENT:")
print("""As per the law, it is a crime for an American to sell
weapons to hostile nations. Country A, an enemy of America,
has some missiles, and all the missiles were sold to it by
Robert, who is an American citizen.""")

# Step 2: Initialize Knowledge Base
KB = {
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
}

# Step 3: Forward Chaining
def forward_chaining(KB):

    new_fact = True

    while new_fact:
        new_fact = False
        inferred = set()

        # Rule 1: Missile(x) -> Weapon(x)
        for fact in KB:
            if fact[0] == "Missile":
                x = fact[1]
                inferred.add(("Weapon", x))

        # Rule 2: Missile(x) AND Owns(A,x)
        # -> Sells(Robert,x,A)
        for m in KB:
            if m[0] == "Missile":
                x = m[1]

                if ("Owns", "A", x) in KB:
                    inferred.add(("Sells", "Robert", x, "A"))

        # Rule 3: Enemy(x,America) -> Hostile(x)
        for fact in KB:
            if fact[0] == "Enemy" and fact[2] == "America":
                x = fact[1]
                inferred.add(("Hostile", x))

        # Rule 4: American(p) AND Weapon(q)
        # AND Sells(p,q,r) AND Hostile(r)
        # -> Criminal(p)

        for a in KB:
            if a[0] == "American":
                p = a[1]

                for w in KB:
                    if w[0] == "Weapon":
                        q = w[1]

                        for s in KB:
                            if s[0] == "Sells" and s[1] == p and s[2] == q:
                                r = s[3]

                                if ("Hostile", r) in KB:
                                    inferred.add(("Criminal", p))

        # Step 4: Add newly inferred facts
        for fact in inferred:
            if fact not in KB:
                KB.add(fact)
                new_fact = True
                print("Inferred:", fact)

    # Step 5: Check the goal
    print("\nGOAL: Criminal(Robert)")

    if ("Criminal", "Robert") in KB:
        print("Robert is a Criminal")
    else:
        print("Cannot prove Robert is a Criminal")


# Step 6: Execute
forward_chaining(KB)

```

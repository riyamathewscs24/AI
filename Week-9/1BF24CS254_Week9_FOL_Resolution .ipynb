```python

# First Order Logic - Resolution

# Step 1: Given Knowledge Base
print("PROBLEM STATEMENT:")
print("""John likes all kinds of food.
Apple and vegetables are food.
Anything anyone eats and is not killed is food.
Anil eats peanuts and is still alive.
Harry eats everything that Anil eats.
Anyone who is alive implies not killed.
Anyone who is not killed implies alive.

Prove: John likes peanuts.""")


# Step 2: Represent facts and rules in CNF

KB = [
    "~Food(x) OR Likes(John,x)",
    "Food(Apple)",
    "Food(Vegetables)",
    "~Eats(x,y) OR Killed(x) OR Food(y)",
    "Eats(Anil,Peanuts)",
    "Alive(Anil)",
    "~Eats(Anil,y) OR Eats(Harry,y)",
    "~Alive(x) OR ~Killed(x)",
    "Killed(x) OR Alive(x)"
]

# Step 3: Negate the goal
goal = "~Likes(John,Peanuts)"

KB.append(goal)

print("\nKnowledge Base in CNF:")
for clause in KB:
    print(clause)

print("\nNegated Goal:", goal)


# Step 4: Apply Resolution

print("\nRESOLUTION STEPS:")

# From Alive(Anil) and
# ~Alive(x) OR ~Killed(x)

print("\n1. Alive(Anil)")
print("   ~Alive(Anil) OR ~Killed(Anil)")
print("   -----------------------------")
print("   ~Killed(Anil)")

# From Eats(Anil,Peanuts) and
# ~Killed(Anil) and
# ~Eats(x,y) OR Killed(x) OR Food(y)

print("\n2. Eats(Anil,Peanuts)")
print("   ~Killed(Anil)")
print("   ~Eats(Anil,Peanuts) OR Killed(Anil) OR Food(Peanuts)")
print("   -----------------------------------------------------")
print("   Food(Peanuts)")

# From Food(Peanuts) and
# ~Food(x) OR Likes(John,x)

print("\n3. Food(Peanuts)")
print("   ~Food(Peanuts) OR Likes(John,Peanuts)")
print("   ---------------------------------------")
print("   Likes(John,Peanuts)")

# Resolve with negated goal

print("\n4. Likes(John,Peanuts)")
print("   ~Likes(John,Peanuts)")
print("   ---------------------")
print("   {}")

# Step 5: Check for empty clause

print("\nEmpty clause obtained!")

print("Goal Proved: John likes peanuts.")
```
